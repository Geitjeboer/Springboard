```python
import sqlite3
```

```python
with sqlite3.connect('sqlite_db_pythonsqlite.db') as conn:
    cursor = conn.cursor()
    cursor.execute("SELECT f.name,  
                   FROM Facilities AS f
                   WHERE Total < 1000
                   ORDER BY Total")
    rows = cursor.fetchall()
    print(rows)
```

```python
with sqlite3.connect('sqlite_db_pythonsqlite.db') as conn:
    cursor = conn.cursor()
    cursor.execute("SELECT b.facid AS fid, b.memid AS mid, b.starttime AS T, Facilities.guestcost*b.slots AS Cost FROM Facilities INNER JOIN Bookings AS b ON b.facid = Facilities.facid UNION SELECT b.facid AS fid, b.memid AS mid, b.starttime AS T, Facilities.membercost*b.slots AS Cost FROM Facilities INNER JOIN Bookings AS b ON b.facid = Facilities.facid") rows = cursor.fetchall()
    print(rows)
    
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import pandas as pd
from PIL import Image, ImageOps
import os

```

```python
#We want to resize all images in the test folders so they are lower resolution.
#We also want to make them gray scale to reduce the number of channels.

def process_images(input_folder, output_folder, target_size):
    """
    resizes all images in greyscale in an input folder and saves them to an output folder.

    Args:
        input_folder (str): Path to the folder containing original images.
        output_folder (str): Path to the folder where processed images will be saved.
        target_size (tuple): A 2-tuple defining the target size (width, height) for resizing.
    """

    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp')):
            input_path = os.path.join(input_folder, filename)
            output_path = os.path.join(output_folder, filename)

            try:
                with Image.open(input_path) as img:
                    # Use ImageOps.grayscale() for a more reliable conversion
                    processed_img = ImageOps.grayscale(img)
                

                    # Resize the processed image
                    resized_img = processed_img.resize(target_size, Image.LANCZOS) # Use LANCZOS for high-quality downsampling

                    # Save the processed image
                    resized_img.save(output_path)
                    print(f"Processed: {filename}")
            except Exception as e:
                print(f"Error processing {filename}: {e}")
```

```python
# process all files in folder list to greyscale and resize to 64x64
cell_types = [
    'basophil',
    'eosinophil',
    'Ig',
    'lymphocyte',
    'monocyte',
    'neutrophil'
]

# Define the target size for resizing (width, height)
target_dimensions = (100, 100)

for folder in cell_types:
    
    input_directory = "..\\Data\\PBC_dataset_normal_DIB_224\\"+folder+"\\"
    output_directory = "..\\Data\\Train_altered\\"+folder+"\\"
    
    process_images(input_directory, output_directory, target_dimensions)
    print("Image processing complete.")
```

```python
#We want to resize all images in the test folders so they only contain the WBC and are lower resolution.
#We also want to keep them RBC.

def process_imagesRBG(input_folder, output_folder, target_size):
    """
    resizes all images in color in an input folder and saves them to an output folder.

    Args:
        input_folder (str): Path to the folder containing original images.
        output_folder (str): Path to the folder where processed images will be saved.
        target_size (tuple): A 2-tuple defining the target size (width, height) for resizing.
    """

    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp')):
            input_path = os.path.join(input_folder, filename)
            output_path = os.path.join(output_folder, filename)

            try:
                with Image.open(input_path) as img:
                    # Use ImageOps.grayscale() for a more reliable conversion
                    processed_img = img.convert('RGB')
                

                    # Resize the cropped image
                    resized_img = processed_img.resize(target_size, Image.LANCZOS) # Use LANCZOS for high-quality downsampling

                    # Save the processed image
                    resized_img.save(output_path)
                    print(f"Processed: {filename}")
            except Exception as e:
                print(f"Error processing {filename}: {e}")

```

All of our WBC images are resized to color and grey scale

```python
# also want to make a copy of colored images cropped and resized

# process all files in folder list
cell_types = [
    'basophil',
    'eosinophil',
    'Ig',
    'lymphocyte',
    'monocyte',
    'neutrophil'
]

# Define the target size for resizing (width, height)
target_dimensions = (100, 100)

# --- Run the processing ---
for folder in cell_types:
    
    input_directory = "..\\Data\\PBC_dataset_normal_DIB_224\\"+folder+"\\"
    output_directory = "..\\Data\\Train_alteredRBG\\"+folder+"\\"

    process_imagesRBG(input_directory, output_directory, target_dimensions)
    print("Image processing complete.")
```

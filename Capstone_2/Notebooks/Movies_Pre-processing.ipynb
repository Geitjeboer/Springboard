```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# Open the Movies_EDA.csv file from the data folder
movies = pd.read_csv('../data/Movies_EDA.csv')
movies.head()
```

```python
# drop the summary and title columns
movies = movies.drop(columns=['Summary', 'Title'])
```

```python
# Change 'Release Date' to only have the month (1-12)

movies['Release Date'] = pd.to_datetime(movies['Release Date']).dt.month
```

```python
movies.head()
```

```python
# Create dummy variables for 'MPAA Rating' and 'Genre'
movies_dummies = pd.get_dummies(movies, columns=['MPAA Rating', 'Genre'],dtype='int', drop_first=False)
```

```python
movies_dummies.head()
```

```python
#Scaling numerical columns but not gross, ROI, or Net Profit becasue it is the target variable
# and we do not want to scale the target variable.
# Columns to scale
scale_cols = ['Budget', 'Runtime', 'Rating', 'Rating Count']

# Initialize scaler and fit_transform
scaler = StandardScaler()
movies_dummies = movies_dummies.copy()
movies_dummies[scale_cols] = scaler.fit_transform(movies_dummies[scale_cols])
```

```python
movies_dummies.head()
```

```python

movies_dummies.shape
```

```python
# Spliting the dataset into training and testing sets
# Predicting 'Net Profit' as the target variable
X = movies_dummies.drop('Net Profit', axis=1)
y = movies_dummies['Net Profit']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
#saving the processed data to a new CSV file
movies_dummies.to_csv('../data/movies_scaled_dummies.csv', index=False)
```

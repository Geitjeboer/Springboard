```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# Open the Movies_EDA.csv file from the data folder
movies = pd.read_csv('../data/Movies_EDA.csv')
movies.head()
```

```python
# Create dummy variables for 'MPAA Rating' and 'Genre'
movies_dummies = pd.get_dummies(movies, columns=['MPAA Rating', 'Genre'],dtype='int', drop_first=False)
```

```python

movies_dummies.head()
```

```python


#Scaling numerical columns
# Columns to scale
scale_cols = ['Budget', 'Gross', 'Runtime', 'Rating', 'Rating Count', 'Net Profit', 'ROI']

# Initialize scaler and fit_transform
scaler = StandardScaler()
movies_scaled = movies.copy()
movies_scaled[scale_cols] = scaler.fit_transform(movies[scale_cols])
```

```python
movies_scaled.head()
```

```python
# Spliting the dataset into training and testing sets
# Predicting 'Net Profit' as the target variable
X = movies_scaled.drop('Net Profit', axis=1)
y = movies_scaled['Net Profit']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
movies_scaled.to_csv('../data/movies_scaled_dummies.csv', index=False)
```
